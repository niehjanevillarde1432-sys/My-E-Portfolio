# What is Regex?

Regex stands for Regular Expressions.
It’s a pattern-matching language used to search, match, and manipulate text.

    Think of it like a "super-powerful find & replace tool."
Instead of just looking for "dog", regex can look for:

- Any digit (\d)

- Any word (\w+)

- An email (\w+@\w+\.\w+)

- A phone number (\d{3}-\d{3}-\d{4})

Why use Regex?

    - Validation → Check if input is a valid email, phone, zip code
    - Search → Find patterns in text (all numbers, dates, hashtags, etc.)
    - Text cleaning → Remove punctuation, extract words, filter unwanted stuff

```python
from typing_extensions import Text
import re

text = "My email is test123@example.com and phone is 123-456-7890."

# 1. Find all digits
print(re.findall(r'\d+' , text))


# 2. Find all words
print(re.findall(r'\w+' , text))


# 3. Find email pattern
print(re.findall(r'\w+@\w+\.\w+' , text))


# 4. Find phone number pattern
print(re.findall(r'\d{3}-\d{3}-\d{4}' , text))



```

# Summary

- Regex helps with text cleaning in NLP:

- Remove punctuation → **re.sub(r"[^\w\s]", "", text)**

- Remove numbers → **re.sub(r"\d+", "", text)**

- Extract URLs → **re.findall(r"http\S+", text)**

- Extract emails → **re.findall(r"...", text)**

- Extract hashtags → **re.findall(r"#\w+", text)**

- Tokenize text → **re.findall(r"\b\w+\b", text)**

- Remove emojis/non-ASCII → **re.sub(r"[^\x00-\x7F]+", "", text)**

## EXAMPLE

```python
#Another approach to stopwords

import re

STOPWORDS = {"is", "am", "are", "the", "a", "an", "and", "at", "in", "on", "for", "of",
             "to", "it", "this"}

#Build regex pattern from stopword list
pattern = r"\b(" + "|".join(STOPWORDS) + r")\b"

# Replace stopwords with empty string
cleaned = re.sub(pattern, "", text, flags=re.IGNORECASE)

# Remove extra spaces
cleaned = re.sub(r"\s+", " ", cleaned).strip()

# ---------------- Example ----------------
text = "This is an example of a sentence in the NLP challenge."
print("Original:", text)
print("No stopwords:", cleaned)
```

# How it works:

- \b → word boundary ensures whole words only.

- Without it, "this" would remove "thistle".

- (" + "|".join(STOPWORDS) + ") → builds a regex group like \b(is|am|are|the|a)\b.

- re.sub() → replaces stopwords with "".

- Clean up extra spaces.
